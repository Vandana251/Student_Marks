Data Cleaning

```python
#Removes duplicate rows from the DataFrame and keeps only unique records.
df = df.drop_duplicates()
```

```python
#Calculates the average attendance percentage of all students.
df["AttendancePercent"].mean()
```

```python
#Saves the cleaned DataFrame as a CSV file without including the DataFrame index.
df.to_csv("cleaned_eda_sample_dataset.csv", index=False)

print("Dataset cleaned successfully!")
```

EDA Process

```python
import matplotlib.pyplot as plt
#Creates a scatter plot to show the relationship between Study Hours and Marks for each student.
plt.scatter(df["StudyHours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
#plt.hist() – Displays the distribution of students' marks by grouping the marks into ranges.
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```
